# OUTLIERS HANDLING

Outliers in AGE, HbA1c, and BMI were retained because they may represent genuine patient characteristics or clinically meaningful observations. Extreme HbA1c and BMI values can provide important information about diabetes severity and associated risk factors, while extreme ages may represent valid patient demographics. Therefore, removing these observations could result in loss of meaningful information.

TASK:
Filter Outliers: In the Creatinine Ratio (Cr) column, exclude values that exceed the
99.5th percentile threshold. Similarly, in the Urea column, exclude values that exceed
the 99.9th percentile threshold. Use these specific percentile thresholds for filtering,
and do not apply IQR or other methods.

```python
cr_threshold = df['Cr'].quantile(0.995)
urea_threshold = df['Urea'].quantile(0.999)
print("Cr 99.5th percentile:", cr_threshold)
print("Urea 99.9th percentile:", urea_threshold)
df = df[
    (df['Cr'] <= cr_threshold) &
    (df['Urea'] <= urea_threshold)
]
print("Cr values above threshold:", (df['Cr'] > cr_threshold).sum())
print("Urea values above threshold:", (df['Urea'] > urea_threshold).sum())
```

# **Task:**
Remove Extreme Outliers: For lipid-related columns (LDL, VLDL, HDL, TG, Chol), remove
extreme outliers using an appropriate method such as IQR or Z-score.

```python
lipid_columns = ['LDL', 'VLDL', 'HDL', 'TG', 'Chol']
for col in lipid_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    df = df[
        (df[col] >= lower_bound) &
        (df[col] <= upper_bound)
    ]

print("Shape after removing lipid outliers:", df.shape)
```

## **Feature Scaling: Standardize or normalize the numerical columns that have varying**
scales (e.g., Patients, AGE, BMI, Cr) to ensure consistent contribution in subsequent
analyses. Justify the scaling method used (standardization vs. normalization).

```python
numeric_columns = df.select_dtypes(include='number').columns
print(df[numeric_columns].agg(['min', 'max']).T)
print(df[numeric_columns].describe().T[['mean', 'std', 'min', 'max']].round(2))
df.select_dtypes(include='number').columns.tolist()
```

```python
from sklearn.preprocessing import StandardScaler
numeric_features = [
    'AGE',
    'Urea',
    'Cr',
    'HbA1c',
    'Chol',
    'TG',
    'HDL',
    'LDL',
    'VLDL',
    'BMI'
]
scaler = StandardScaler()
df[numeric_features] = scaler.fit_transform(df[numeric_features])
print(df[numeric_features].describe().round(2))
```

Standardization was chosen because the clinical numerical variables have different ranges and units. For example, Cr ranges from 6 to 401, while HDL ranges from 0.4 to 1.9. StandardScaler transforms the variables to a common scale with a mean of approximately 0 and a standard deviation of 1. Identifier columns such as Unnamed: 0, Visit_ID, and Patient_id were excluded because their numerical values represent identifiers rather than meaningful measurements
